```python
import pandas as pd
import numpy as np
import re

df_a = pd.read_csv('/content/drive/MyDrive/Week 1/Dataset/transactions_dump_partA.csv', low_memory=False)
df = df_a.copy()


#Clean store_id

def normalize_store_id(x):
    if pd.isna(x):
        return np.nan
    x = str(x).upper()
    x = re.sub(r'[^A-Z0-9]', '', x)
    m = re.search(r'S(\d+)', x)
    if m:
        return f"S{m.group(1).zfill(2)}"
    return np.nan

df["store_id_clean"] = df["store_id"].apply(normalize_store_id)

#Clean product_category
def clean_category(x):
    if pd.isna(x):
        return np.nan
    x = str(x).lower().strip()
    corrections = {"grocry": "grocery", "electronix": "electronics"}
    return corrections.get(x, x)

df["product_category_clean"] = df["product_category"].apply(clean_category)


#Clean unit_price
def parse_price(x):
    if pd.isna(x):
        return np.nan
    x = str(x).lower().strip()
    if x == "free":
        return 0.0
    x = re.sub(r'[^\d\.]', '', x)
    try:
        return float(x)
    except:
        return np.nan

df["unit_price_clean"] = df["unit_price"].apply(parse_price)

#Clean quantity
word_to_num = {"zero":0,"one":1,"two":2,"three":3,"four":4,"five":5,"six":6,"seven":7,"eight":8,"nine":9,"ten":10}

def parse_quantity(x):
    if pd.isna(x):
        return np.nan
    x = str(x).lower().strip()
    if x in word_to_num:
        return word_to_num[x]
    try:
        return float(x)
    except:
        return np.nan

df["quantity_clean"] = df["quantity"].apply(parse_quantity)

#Clean discount
def parse_discount(x):
    if pd.isna(x):
        return np.nan
    x = str(x).strip().replace('%', '')
    try:
        return float(x)
    except:
        return np.nan

df["discount_clean"] = df["discount"].apply(parse_discount)


#Clean sales_channel
def clean_channel(x):
    if pd.isna(x):
        return np.nan
    x = str(x).lower().strip()
    if x in ["online", "offline"]:
        return x
    return np.nan

df["sales_channel_clean"] = df["sales_channel"].apply(clean_channel)

#Clean transaction_date
df["transaction_date_clean"] = pd.to_datetime(df["transaction_date"], errors="coerce")

# Assign data tiers
def assign_tier(row):
    if row["quantity_clean"] < 0:
        return "TIER_3_IMPOSSIBLE"
    if row["unit_price_clean"] < 0:
        return "TIER_3_IMPOSSIBLE"
    if pd.notna(row["unit_price_clean"]) and pd.notna(row["discount_clean"]) and row["discount_clean"] > row["unit_price_clean"]:
        return "TIER_3_IMPOSSIBLE"
    if pd.notna(row["transaction_date"]) and pd.isna(row["transaction_date_clean"]):
        return "TIER_3_IMPOSSIBLE"

    if pd.isna(row["quantity_clean"]) or pd.isna(row["unit_price_clean"]) or pd.isna(row["transaction_date_clean"]):
        return "TIER_2_SUSPICIOUS"

    return "TIER_1_CLEAN"

df["data_tier"] = df.apply(assign_tier, axis=1)

#Final cleaned dataframe
df_a_cleaned = df[df["data_tier"].isin(["TIER_1_CLEAN", "TIER_2_SUSPICIOUS"])][
    ["invoice_id", "store_id_clean", "product_category_clean",
     "unit_price_clean", "quantity_clean", "transaction_date_clean",
     "discount_clean", "sales_channel_clean", "data_tier"]
].copy()

#Report

print("Tier distribution:")
print(df["data_tier"].value_counts())

print("\nRows before:", len(df))
print("Rows after keeping Tier 1 & 2:", len(df_a_cleaned))
```

```python
import pandas as pd
import numpy as np
import re

# Load df_c
df_c = pd.read_csv("/content/drive/MyDrive/Week 1/Dataset/store_master_corrupt.csv", low_memory=False)
dfc = df_c.copy()

# 1. Improved store_id normalization
def normalize_store_id(x):
    if pd.isna(x):
        return np.nan
    x = str(x).upper()
    x = re.sub(r'[^A-Z0-9]', '', x)
    x = x.replace("STORE", "")  # remove literal STORE
    m = re.search(r'(\d+)', x)
    if m:
        return f"S{m.group(1).zfill(2)}"
    return np.nan

dfc["store_id_clean"] = dfc["store_id"].apply(normalize_store_id)


# 2. Clean city

def clean_city(x):
    if pd.isna(x):
        return np.nan
    x = str(x).lower().strip()
    corrections = {"ahemdabad": "ahmedabad"}
    return corrections.get(x, x)

dfc["city_clean"] = dfc["city"].apply(clean_city)

# 3. Clean store_type
def clean_store_type(x):
    if pd.isna(x):
        return np.nan
    return str(x).lower().strip()

dfc["store_type_clean"] = dfc["store_type"].apply(clean_store_type)

# ----------------------------
# 4. Clean opening_year
# ----------------------------
word_years = {
    "two thousand thirteen": 2013,
    "two thousand twelve": 2012,
    "two thousand ten": 2010
}

def parse_year(x):
    if pd.isna(x):
        return np.nan
    x = str(x).lower().strip()
    if x in word_years:
        return word_years[x]
    m = re.search(r'(19|20)\d{2}', x)
    if m:
        return int(m.group())
    try:
        return int(x)
    except:
        return np.nan

dfc["opening_year_clean"] = dfc["opening_year"].apply(parse_year)


#Final df_c_cleaned (no dropping)
df_c_cleaned = dfc.copy()

# 6. Report
print("df_c_cleaned:")
print(df_c_cleaned)

print("\nMissing %:")
print(df_c_cleaned[["store_id_clean", "city_clean", "store_type_clean", "opening_year_clean"]].isna().mean() * 100)
```

```python
# Make sure we're working on a real copy (avoid SettingWithCopyWarning)
model_df = model_df.copy()

# 1. Create month feature
model_df.loc[:, "month"] = model_df["transaction_date_clean"].dt.month

# 2. Compute discount mode from Tier-1 only
discount_mode = (
    model_df
    .loc[(model_df["data_tier"] == "TIER_1_CLEAN") & model_df["discount_clean"].notna()]
    .groupby(["product_category_clean", "month"])["discount_clean"]
    .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
)

# 3. Impute discount into Tier-2 only
mask = (
    (model_df["data_tier"] == "TIER_2_SUSPICIOUS") &
    (model_df["discount_clean"].isna())
)

model_df.loc[mask, "discount_clean"] = (
    model_df.loc[mask]
    .set_index(["product_category_clean", "month"])
    .index.map(discount_mode)
)

# 4. Report effect
print("Missing discount after imputation (%):", model_df["discount_clean"].isna().mean() * 100)
print("Imputed rows:", mask.sum())
```

Title: Recovering Truth from Broken Retail Transaction Data
## 1. Introduction

The provided retail datasets represented transactional logs from legacy systems that had undergone a failed ERP migration. The data was highly fragmented, inconsistent, and partially corrupted. Our objective was not only to build a machine-learning model but first to determine whether the data was trustworthy enough for analytics at all.

Instead of applying naïve cleaning, we designed a trust-aware data recovery pipeline that:

Aligns heterogeneous schemas

Converts corrupted numerical and categorical fields

Detects logical contradictions

Classifies rows by reliability

Preserves uncertainty instead of hiding it

Three datasets were provided:

transactions_dump_partA.csv – POS system A

transactions_dump_partB.csv – POS system B with schema mismatch

store_master_corrupt.csv – store metadata with textual corruption

##2. Intelligent Data Ingestion

The first challenge was that the two transaction files used different schemas and inconsistent representations:

Column names differed (qty vs quantity, price vs unit_price)

Mixed delimiters and formats

Currency symbols inside numeric fields

Dates in more than five formats

Text values such as “ten” instead of numbers

We performed:

Manual schema alignment

Canonical column naming

Safe type conversions with error handling

Explicit documentation of information loss

The store master file contained additional issues:

Multiple spellings for the same store

Textual years such as “two thousand thirteen”

Missing city and store type

This file was treated as metadata, not ground truth.

##3. Data Cleaning & Preprocessing
###3.1 Safe Conversion of Corrupted Fields

Custom parsers were built:

parse_price() removed ₹/$ symbols and converted “free” → 0

parse_quantity() converted words (“ten”) to numbers

clean_category() standardized spellings (grocry → grocery)

clean_channel() normalized online/offline

Robust date parsing with coercion

###3.2 Logical Validation

Business rules were enforced:

quantity < 0 → invalid

price < 0 → invalid

discount > price → impossible

unparseable dates → unreliable

###3.3 Tier-Based Trust Model

Instead of deleting everything, each row was classified:

Tier-1 (Clean): All core fields valid and consistent

Tier-2 (Suspicious): No contradictions but incomplete

Tier-3 (Impossible): Violates business logic

Results:

Dataset	Tier-1	Tier-2	Tier-3
df_a	~14.7k	~14.8k	~170k
df_b	~17.7k	~15.5k	~166k

Only ~15% of raw data was truly reliable — a major analytical finding in itself.

##4. Conditional Data Recovery
###4.1 Discount Imputation

Discount was 66% missing.
A naïve fill would inject false information, so we:

Learned patterns only from Tier-1

Grouped by (product_category × month)

Used statistical mode

Filled only Tier-2 rows

This recovered 5,574 values, reducing missingness to 44% while preserving honesty.

###4.2 Store Master Handling

Store IDs were normalized

City and type standardized

Conflicting records preserved rather than collapsed

The merged dataset was deduplicated to avoid artificial inflation.

##5. Final Dataset Characteristics

After recovery:

~62,860 unique transactions

Reliable fields:

transaction date

unit price

partially discount

Unreliable: store, channel, category

This guided feature selection and modeling strategy.

##6. Key Insights from Cleaning

The majority of logs were system-generated noise

Missingness was structured, not random

Discount behavior followed seasonal patterns

Store metadata was too inconsistent for causal modeling

###1. Introduction

This project dealt with a real–world scenario where retail transactional data became heavily corrupted after a failed ERP migration. The organization required quick analytical insights such as factors influencing sales and the possibility of building a machine-learning baseline. However, the raw data was fragmented, inconsistent, and logically unreliable.

The objective therefore became twofold:

Recover maximum truthful information from corrupted CSV logs.

Evaluate whether meaningful machine learning predictions are possible from the recovered data.

Three datasets were provided:

transactions_dump_partA.csv – POS system A

transactions_dump_partB.csv – POS system B with schema mismatch

store_master_corrupt.csv – store metadata with textual corruption

###2. Data Challenges

Initial inspection revealed severe quality issues:

Mixed column names across files

Numeric fields stored as text (“ten”, “free”, ₹124.29)

Dates in multiple formats and many “invalid_date” values

Negative quantities, discount greater than price

40–60% missing values in key columns

Duplicate invoices and category spelling noise

Because of these problems, no row could be trusted blindly.

###3. Intelligent Data Ingestion

The two transaction files were aligned into a common schema. Custom parsers were developed:

parse_price() removed currency symbols and converted “free” → 0

parse_quantity() converted word numbers to integers

clean_category() standardized noisy labels

clean_channel() normalized online/offline

Robust date parsing with coercion

The store master file contained additional problems such as textual years (“two thousand thirteen”) and inconsistent city names. It was treated as auxiliary metadata rather than ground truth.

###4. Tier-Based Data Validation

To avoid either blind deletion or blind acceptance, a trust model was introduced.

Tier Definitions

Tier-1 (Clean): logically consistent and fully usable

Tier-2 (Suspicious): incomplete but not contradictory

Tier-3 (Impossible): violates business logic (e.g., negative quantity, discount > price)

Results
Dataset	Tier-1	Tier-2	Tier-3
Part A	~14k	~14k	~170k
Part B	~17k	~15k	~166k

Only about 15% of the raw logs were truly reliable, which itself became a major project insight.

###5. Conditional Data Recovery
5.1 Discount Imputation

Discount was 66% missing. Instead of mean filling, a controlled strategy was used:

Learn patterns only from Tier-1

Group by (product_category × month)

Use statistical mode

Fill only Tier-2 rows

This recovered 5,574 values, reducing missingness to ~44% without fabricating information.

####5.2 Store ID Assumption and Correction

Initially, store IDs such as store_1, S-001, Store-01 were assumed to be formatting variants and were normalized to a single ID.
Later inspection revealed this was conceptually wrong—they were distinct stores.

The pipeline was corrected by:

Rolling back normalization

Using raw store identifiers

Re-merging metadata via exact match only

This revision was documented as a learning step demonstrating iterative understanding.

###6. Exploratory Data Analysis
####6.1 Effect of Log Transformation

Raw distributions of price and quantity were extremely right-skewed.
Boxplots appeared empty due to scale compression.
A log(1+x) transformation was applied:

Made central tendency visible

Allowed identification of genuine extreme transactions

Stabilized variance for modeling

####6.2 Relationship Analysis

Scatter of price vs quantity showed weak inverse trend

Discount vs quantity showed mild positive association

Correlation heatmap revealed near-zero linear correlations, indicating non-linear behavior

####6.3 Business Question Findings

Store/channel analysis could not be trusted due to missing metadata
